## Matrix Math


### Scalar Operations


```python
n = 1
a = np.array([1, 2, 2])
print("Scalar Operations")
print(f"n = {n}, scalar")
print(f"a = {a}, shape: {a.shape}")
print(f"a + n: {a + n}")
```

### Vector Addition/Subtraction


```python
a = np.array([1, 2, 3])
b = np.array([3, 5, 6])
print("Vector Addition/Subtraction")
print(f"a = {a}, shape: {a.shape}")
print(f"b = {b}, shape: {b.shape}")
print(f"a + b = {a + b}")
```

### Dot Product


```python
a = np.array([1, 2, 3])
b = np.array([4, -2, 1])
print("Dot Product")
print(f"a = {a}, shape: {a.shape}")
print(f"b = {b}, shape: {b.shape}")
print(f"a dot b = {a @ b}")
```

### Matrix Transposition


```python
A = np.array([[1, 2], [3, -4], [0.5, 2]])
print("Matrix Transpose")
print(f"A = {A}, shape: {A.shape}")
print(f"matrix transpose: {A.T}, shape: {A.T.shape}")
```

### Matrix Addition/Subtraction


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, -2], [1, 5]])
print("Matrix Addition")
print(f"A = {A}, shape: {A.shape}")
print(f"B = {B}, shape: {B.shape}")
print(f"A + B = {A + B}")
```

### Matrix Scalar Operations

```python
n = 2
A = np.array([[1, 2], [3, -4], [0.5, 2]])
print("Matrix Scalar Operations")
print(f"n = {n}, scalar")
print(f"A = {A}, shape: {A.shape}")
print(f"nA = {n * A}")
```

### Matrix Multiplication


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, -2], [1, 5]])
print("Matrix Multiplication")
print(f"A = {A}, shape: {B.shape}")
print(f"B = {B}, shape: {B.shape}")
print(f"A mult B = {A @ B}")
```
