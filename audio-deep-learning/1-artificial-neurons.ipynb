## Artificial Neurons


```python
def sigmoid(x):
    """
    Activation Function - Sigmoid is very common. there are many others like RELU, TANH etc
    Activation functions should be differentiable continuous functions because we need to take the gradient

    Args:
        x(float): input value to be processed
    Returns:
        y(float): output of the function
    """
    return 1.0 / (1 + math.exp(-x))
```

```python
def activate(inputs, weights):
    """
    Computes the activation of a neuron based on input signals and connection weights.
    Output = f(x_1*w1 + x_2*w2 + .... + x_k*wk) where f is the sigmoid function (in this case)

    Args:
        inputs(list): input signals
        weights(list): connections to weights
    Returns:
        output(float): output value

    """
    # net input
    h = 0
    for x, w in zip(inputs, weights):
        h += x * w

    # activation
    return sigmoid(h)
```
