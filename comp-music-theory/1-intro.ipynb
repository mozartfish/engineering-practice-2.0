## Bach Chorale Analysis

```python
bach = corpus.parse("bach/bwv66.6")
bach.show()
```

### Alto Part 

```python
bach.parts["alto"].show()
```

### Soprano Part 

```python
sopr = bach.parts[0]
```

### Soprano Part - Measures 3-5(Inclusive)

```python
sopr.measures(3, 5).show()
```

### Manipulate Score 

```python
c = sopr.measure(5).notes[2]  # technically a c#
```

```python
c.style.color = "red"
```

```python
sopr.show()
```

### Key Signature

```python
bach.analyze("key")
```

### Key Signature for Measures 0-2 (Inclusive)

```python
bach.measures(0, 2).analyze("key")
```
